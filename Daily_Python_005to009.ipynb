### **005**：去除列表的重复元素

```python
# 使用for循环：保留原顺序，也就是第一次出现的顺序
list_ = ["Python", "C", "Java", "C", "MATLAB", "Python", "Python"]
print("0：", list_)

list_revised = []    # 初始化
print("1：", list_revised)
for item in list_:    # 遍历每个元素
    if item in list_revised:    # 如果元素item已经在列表list_revised中，则应跳过本次循环
        continue
    list_revised.append(item)
print("2：", list_revised)

list_revised = []    # 初始化
print("3：", list_revised)
for item in list_:    # 遍历每个元素
    if item not in list_revised:    # 如果元素item不在在列表list_revised中，则执行添加远不步骤
        list_revised.append(item)
print("4：", list_revised)
```

```python
# 使用dict.fromkeys()，将列表元素转换成字典的键：保留原顺序
list_ = ["Python", "MATLAB", "C", "Java", "Rust", "C", "MATLAB", "Python", "Python"]
print("0：", list_)

# 将元素转换成字典的键，因为键不会有重复的，重复出现只会覆盖原有的
dictionary = dict.fromkeys(list_)    # 列表元素转换成字典的键
print("1：", type(dictionary))
print("2：", dictionary)

list_revised = list(dictionary)    # 字典的键转换成列表
print("3：", list_revised)
```

```python
# 使用set()，将列表元素转换为集合的元素：不保证原顺序，因为set()是根据哈希表来排序的
list_ = ["Rust", "MATLAB", "C", "Java", "C", "MATLAB", "Python", "Python"]
print("0：", list_)

set_ = set(list_)    # 列表转换成集合
print("1：", type(set_))
print("2：", set_)

list_revised = list(set_)    # 集合转换成列表
print("3：", list_revised)
```

### **006**：给列表元素排序

```python
# 用sort()，直接修改原列表
list_ = [3, 2, 4, 1, 0]
print("0：", list_)

list_.sort()    # 升序排列，直接修改原列表
print("1：", list_)

list_.sort(reverse=True)    # 降序排列，直接修改原列表
print("2：", list_)
```

```python
# 用sorted()，不会修改原列表
list_ = [3, 2, 4, 1, 0]
print("0：", list_)

list_revised = sorted(list_)    # 升序排列，不会修改原列表，需要返回给新变量
print("1：", list_revised)

list_revised = sorted(list_revised, reverse=True)    # 降序排列，直接修改原列表
print("2：", list_revised)
```

```python
# 用key，自己指定依据
list_ = ["apple", "orange", "pear"]
print("0：", list_)

list_revised = sorted(list_, key=len)    # 根据字符串长度排列，升序
print("1：", list_revised)

print("2：", list_)

list_.sort(key=len, reverse=True)    # 按照元组中第1个元素排序，升序。x表示列表的元素，可换成其他符号
print("3：", list_)
```

```python
# 用key，处理复杂元素的列表
list_ = [(2, "b"), (1, "a"), (3, "c")]    # 列表元素为元组
print("0：", list_)

list_revised = sorted(
    list_,
    key=lambda x: x[0],    # 按照元组的第1个元素排列
    reverse=True    # 降序排列
)
print("1：", list_revised)

print("2：", list_)

list_.sort(key=lambda x: x[0])    # 按照元组中第1个元素排序，升序。x表示列表的元素，可换成其他符号
print("3：", list_)
```

### **007**：统计列表元素出现次数

```python
# 用字典+get()
list_ = ["Python", "Rust", "MATLAB", "C", "Java", "C", "MATLAB", "Python", "Python"]
print("0：", list_)

counts = {}    # 初始化，用于将列表元素转换成字典的键
for item in list_:    # 遍历列表的每一个元素
    counts[item] = counts.get(item, 0) + 1    # get()获取字典counts的键item对应的值，若该键不存在则返回“0”
    """
    等式左边：将某一值赋值给字典counts的键item
    等式右边：若item没在字典中，表明字典还没统计该元素，则右边得到0+1==1，
             若item已经在字典中，表明字典已经统计过该元素次数，用n表示次数，再加上这次，则右边得到n+1
    """
print("1：", counts)
```

```python
# 用count()：需要指定要统计次数的元素
list_ = ["Python", "Rust", "MATLAB", "C", "Java", "C", "MATLAB", "Python", "Python"]
print("0：", list_)

number0 = list_.count("Python")    # 统计元素"Python"出现的次数
print("1：", number0)
```

```python
# 使用Counter()：一次统计所有元素出现的次数
from collections import Counter    # 这是Python标准库里面的，不需要额外下载

list_ = ["Python", "Rust", "MATLAB", "C", "Java", "C", "MATLAB", "Python", "Python"]
print("0：", list_)

counts = Counter(list_)
print("1：", type(counts))    # 类型为：<class 'collections.Counter'>
print("2：", counts)

counts = dict(counts)    # 转换成普通字典
print("3：", type(counts))
print("4：", counts)
```

## **008**：拼接列表、字典

```python
# 拼接不同的列表
list_0 = [1, 2, 3]
list_1 = ["Python", "C"]

list_2 = list_0 + list_1    # 列表相加，就是列表拼接
print(f"拼接后：{list_2}")

# 拼接相同的列表
list_3 = list_0 * 3    # 列表重复拼接3次
print(f"重复拼接3次：{list_3}")
```

```python
# 拼接不同的字典：直接修改原字典，覆盖掉原字典的同名键
dictionary_0 = {
    "Python": 0,
    "C": 1
}
dictionary_1 = {
    "Python": 3,
    "C++": 2
}
print(f"字典0：{dictionary_0}")
print(f"字典1：{dictionary_1}")

dictionary_0.update(dictionary_1)
print(f"更新后字典0：{dictionary_0}")    # 键'Python'的值被覆盖，添加了键值对'C++':2
print(f"更新后字典1：{dictionary_1}")    # 无变化
```
