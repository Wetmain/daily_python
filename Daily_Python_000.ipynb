### **000**：列表的用法

```python
# 直接创建列表
list_0 = [1, 2, 3]    # 元素类型：int
print("0：", type(list_0))

list_1 = [1.2, 2.0, 3]    # 元素类型：float + int
print("\n1：", type(list_1))
print("1：", list_1)

list_2 = [1, "Python", True]    # 元素类型：int + str + Boolean
print("\n2：", type(list_2))
print("2：", list_2)

list_3 = [[1, 2], [2, 3]]    # 元素类型：list
print("\n3：", type(list_3))

list_4 = [{1: 2}, {2: 3, 3: 4}]    # 元素类型：dict
print("\n4：", type(list_4))

list_5 = [(1, 2), (2, 3)]    # 元素类型：tuple
print("\n5：", type(list_3))
```

```python
# 用list()，将可迭代对象转换成列表
# 将元组转换成列表
a = (1, 2, 3)
print("0：", type(a))
print("0：元组", a)
list_0 = list(a)    # 转换成列表
print("0：", type(list_0))
print("0：列表", list_0)

# 将字符串拆分后转换成列表
b = "python"
print("\n1：", type(b))
print("1：字符串", b)
list_1 = list(b)    # 注意，字符串的每一部分各自变成列表的元素
print("1：", type(list_1))
print("1：列表", list_1)    # ['p', 'y', 't', 'h', 'o', 'n']

# 将惰性序列转换成列表
c = range(5)    # 左闭右开
print("\n2：", type(c))
print("2：惰性序列", c)
list_2 = list(c)    # 转换成列表
print("2：", type(list_2))
print("2：列表", list_2)

# 将字典的键转换成列表
d = {1:2, "s": 0}
print("\n3：", type(c))
print("3：字典", d)
list_3 = list(d)    # 转换成列表
print("3：", type(list_3))
print("3：列表", list_3)
```

```python
# 用列表推导式创建列表
# 创建[0, 1, 4, 9, 16, 25]
list_ = [i ** 2 for i in range(6)]
print("0：", list_)

# 上面的执行过程
list_revised = []
print("1：", list_revised)

for i in range(6):    # 遍历列表的每个元素
    list_revised.append(i ** 2)
print("2：", list_revised)



```

```python
# 列表推导式：添加筛选条件
list_ = list(range(6))    # 创建[0, 1, 2, 3, 4, 5]
print("0：", list_)

list_revised = [i for i in list_ if i % 2 == 0]    # 筛选条件：能被2整除
print("1：", list_revised)

# 多个筛选条件
list_revised = ["偶数" if i % 2 ==0 else "奇数" for i in list_]
"""
原列表元素能被2整数，新列表对应元素就为"偶数"，否则为"奇数"
"""
print("2：", list_revised)
```

```python
# 列表推导式：筛选元素部分内容
list_ = [(3, "c"), (1, "a"), (2, "b")]
print("0：", list_)

list_revised = [i[0] for i in list_]    # 筛选列表元素的第1个内部元素
print("1：", list_revised)

list_revised = [i[1] for i in list_]    # 筛选列表元素的第2个内部元素
print("1：", list_revised)
```

```python
# 通过索引获取单个元素
list_ = [1, 2, 3, 4, 5, 6]
print(list_[0])    # 第1个元素：1
print(list_[3])    # 第4个元素：4
print(list_[-1])    # 最后1个元素：6
print(list_[-2])    # 倒数第2个元素：5
```

```python
# 通过切片获取获取子列表
list_ = [1, 2, 3, 4, 5, 6]
print(list_)
print("获取获取索引1到3个元素的子列表：", list_[1:4])    # 获取获取索引1到3个元素的子列表：[2,3,4]
print("获取索引4到最后元素的子列表：", list_[4:])    # 获取索引4到最后元素的子列表：[5,6]
print("获取开始到索引2元素的子列表：", list_[:3])    # 获取开始到索引2元素的子列表：[1,2,3]
print("获取去除最后1个元素的子列表：", list_[:-1])    # 获取去除最后1个元素的子列表：[1,2,3,4,5]
print("从索引1开始，按步长2取元素的子列表：", list_[1::2])    # 从索引1开始，按步长2取元素的子列表：[2,4,6]

print("反转元素顺序：", list_[::-1])    # 反转元素顺序：[6,5,4,3,2,1]

print("浅拷贝：", list_[:])    # 获取全部元素的列表，注意是浅拷贝
```

```python
# 修改列表元素
list_ = [1, 2, 3, 4, 5, 6]
print(list_)

list_[1] = 9    # 直接通过索引给相应的元素重新赋值，从而修改列表
print("索引1修改为9：", list_)

list_[1:3] = [0, 0]      # 通过切片给多个元素重新赋值，从而修改列表
print("索引1,2修改为0,0：", list_)

list_[2:4] = [7, 9, 8]    # 切片2个元素，赋值为3个元素，从而修改列表并扩展其长度
print(f"索引2,3修改为7,9，并在在索引3后面插入元素8：{list_}")
```

```python
# 插入元素
list_ = [1, 2]
print(f"{list_}")

list_.append("Python")    # 在列表末尾添加1个元素
print(f"列表末尾添加1个元素：{list_}")

list_.append(["C", "MATLAB"])    # 这将作为1和元素，添加至列表末尾
print(f"列表末尾添加1个元素：{list_}")

list_.extend(range(4, 8))    # 在列表末尾逐个添加可迭代对象
print(f"将可迭代对象逐个添加至列表末尾：{list_}")

list_.insert(1, 0)    # 在索引1前面，插入元素0
print(f"在索引1前面，插入元素0:{list_}")
```

```python
# 删除元素
list_ = [1, 2, 3, 4, 5, 5,  1, 1]
print(list_)

list_.remove(1)     # 删去第1个等于1的元素
print(f"删去第一个等于1的元素：{list_}")

list_.pop()    # 删去最后一个元素
print(f"删去最后一个元素：{list_}")

del list_[3]    # 删去索引3的元素
print(f"删去索引3的元素：{list_}")

del list_[1:3]    # 删去索引1,2的元素
print(f"删去索引1,2的元素：{list_}")

list_.clear()    # 清空列表
print(f"清空列表：{list_}")
```

```python
# 列表长度
list_ = [1, 3, 5, 7, 3]
print(list_)
print(f"列表长度：{len(list_)}")

# 返回布尔值
print(f"“1”在列表中：{1 in list_}")
print(f"“3”不在列表中：{3 not in list_}")

# 查找元素在列表的索引
print(f"第一个“3”的索引：{list_.index(3)}")

# 元素数据比较
print(f"所有元素之和：{sum(list_)}")
print(f"元素最大值：{max(list_)}")
print(f"元素最小值：{min(list_)}")
```

```python
# 遍历列表：只需要元素
list_ = ["Python", "C", "Rust", "MATLAB"]
print(list_)
for item in list_:
    print(item)

# 遍历列表：同时获取索引与元素
print("\n")
for order, item in enumerate(list_):
    print(order, "：", item)

# 遍历列表：元素类型为元组
print("\n")
list_ = [("张三", 90), ("李四", 88)]
for name, score in list_:
    print(name, "：", score)
```

```python
# 嵌套列表的索引
list_0 = [
    [1, 2, 3],
    [2, 3, 4]
]
print(f"第1行：{list_0[0]}")   # 第1行，全部列
print(f"第1行第2列：{list_0[0][1]}")    # 第1行第2列

# 嵌套列表的展平
list_1 = [i for row in list_0 for i in row]    # 对于外部列表的每一个内部列表，对于内部列表的每一个元素
print(f"展平：{list_1}")

# 上述操作展开写
list_1 = []
print(list_1)
for row in list_0:    # 遍历外部元素的每一个元素，也就是每一个内部列表
    for i in row:    # 遍历内部列表每一个元素
        list_1.append(i)
print(f"展平：{list_1}")
```
