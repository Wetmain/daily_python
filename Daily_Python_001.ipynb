### **001**：字典的用法

```python
# 直接创建字典
dictionary_0 = {
    "Python": 1,
    "C": 2
}
print(f"类型：{type(dictionary_0)}")
print(f"查看字典：\n{dictionary_0}")
```

```python
# 将二维列表转换成字典：有键和值
list_1 = [("Python", 1), ("C", 2)]    # 元素为元组
dictionary_2 = dict(list_1)
print(f"列表内有元组：{dictionary_2}")

list_2 = [["Python", 2], ["C", 1]]    # 元素为列表
dictionary_3 = dict(list_2)
print(f"列表内有列表：{dictionary_3}")

# 用dict()根据关键字参数
dictionary_4 = dict(a="Python", number=1)
print(dictionary_4)
```

```python
# 用dici()，根据zip()配对两组数据
zip_0 = zip(
    ["Python", "C", "Rust"],
    [4, 2, 1]
)
print(zip_0)

dictionary_0 = dict(zip_0)
print(dictionary_0)
```

```python
# 用fromkeys()
keys = ["Python", "C", "Rust"]

dictionary_0 = dict.fromkeys(
    keys,    # 各个键
    [0, 1, 2]    # 每个键的值（所有键都是这个值）
)
print(dictionary_0)
```

```python
# 将1维列表转换成字典：列表元素作为键
list_0 = ["Python", "C", "Rust"]

dictionary_0 = dict.fromkeys(list_0)    # 键的值为None
print(f"未指定键的值：{dictionary_0}")

dictionary_1 = dict.fromkeys(list_0, 0)    # 键的值为0
print(f"指定键的值为0：{dictionary_1}")
```

```python
# 将1维列表转换成字典：列表元素索引作为键，列表元素作为值
list_0 = ["Python", "C", "Rust"]

a = enumerate(list_0)
print(f"获取索引与元素，并配对：{a}")
dictionary_0 = dict(a)
print(dictionary_0)
```

```python
# 获取字典的“长度”：键值对的数量
dictionary_0 = {
    "Python": 0,
    "C": 1
}
print(f"字典：{dictionary_0}")

length = len(dictionary_0)
print(f"键值对的数量：{length}")
```

```python
# 获取字典的键所对应的值
dictionary_0 = {
    "Python": 0,
    "C": 1
}
print(f"字典：{dictionary_0}")

# 直接查看键的值：没有的键会报错
print(f"\n获取键'Python'对应的值：{dictionary_0["Python"]}")

# 用get()
print(f"\n获取键'C'对应的值：{dictionary_0.get("C")}")
print(f"（未设置默认值）获取不存在的键'MATLAB'对应的值：{dictionary_0.get("MATLAB")}")    # 未设置默认值会返回None
print(f"（设置默认值）获取不存在的键'MATLAB'对应的值：{dictionary_0.get("MATLAB", "没有该键")}")    # 指定默认值："没有该键"
```

```python
# 获取键对应的值，若指定的键不存在，则给字典添加指定的键值对
dictionary_0 = {
    "Python": 0,
    "C": 1
}
print(f"字典：{dictionary_0}")

a = dictionary_0.setdefault("Python", 5)    # 键已存在，则返回其值
print(f"\n已存在的键'Python'的值：{a}")

b = dictionary_0.setdefault("C++", 3)    # 键不存在，则给字典添加键值对"C++":3，并返回默认值3
print(f"不存在的键'C++'，指定默认值3后：{a}")
print(f"新的字典：{dictionary_0}")
```

```python
# 判断键是否存在
dictionary_0 = {
    "Python": 0,
    "C": 1
}
print(f"字典：{dictionary_0}")

print(f"\n存在的键C：{"C" in dictionary_0}")    # True
print(f"不存在的键MATLAB：{"MATLAB" in dictionary_0}")    # False
```

```python
# 判断一个值是否在字典中
dictionary_0 = {
    "Python": 0,
    "C": 1
}
print(f"字典：{dictionary_0}")

print(f"\n存在的值0：{0 in dictionary_0.values()}")    # True
print(f"不存在的值2：{2 in dictionary_0.values()}")    # False
```

```python
# 给键赋值：若键已存在，则修改其值，若键不存在，则添加该键值对
dictionary_0 = {
    "Python": 0,
    "C": 1
}
print(f"字典：{dictionary_0}")

dictionary_0["Python"] = 2    # 键已存在，再赋值则是修改其值
print(f"\n给已存在的键'Python'重新赋值为2：{dictionary_0}")

dictionary_0["C++"] = 3    # 键不存在，赋值则是给字典添加键值对
print(f"给不存在的键'C++'赋值为3：{dictionary_0}")
```

```python
# 删除键值对：用del
dictionary_0 = {
    "Python": 0,
    "C": 1,
    "C++": 2
}
print(f"字典：{dictionary_0}")

# 删除指定键，值自然也会删除。键不存在时会报错
del dictionary_0["C"]
print(f"\n删除键'C'后的字典：{dictionary_0}")
```

```python
# 删除键值对，并返回该值
dictionary_0 = {
    "Python": 0,
    "C": 1,
    "C++": 2,
    "Rust": 3
}
print(f"字典：{dictionary_0}")

# 不指定默认值时，若键不存在也会报错
a = dictionary_0.pop("C")
print(f"\n被删除的键值对的值：{a}")
print(f"删除键'C'后的字典：{dictionary_0}")
```

```python
# 删除键值对，并给出默认值防止报错
dictionary_0 = {
    "Python": 0,
    "C": 1,
    "C++": 2,
    "Rust": 3
}
print(f"字典：{dictionary_0}")

# 指定默认值："不存在的键"
a = dictionary_0.pop("MATLAB", "不存在的键")
print(f"\n想删除不存在的键值对，返回的默认值：{a}")
print(f"字典无变化：{dictionary_0}")
```

```python
# 删除并返回最后的键值对
dictionary_0 = {
    "Python": 0,
    "C": 1,
    "C++": 2,
    "Rust": 3
}
print(f"字典：{dictionary_0}")

# 空字典时会报错
a, b = dictionary_0.popitem()
print(f"被删除的最后的键值对：{(a, b)}")
print(f"变化后的字典：{dictionary_0}")
```

```python
# 清空字典
dictionary_0 = {
    "Python": 0,
    "C": 1,
    "C++": 2,
    "Rust": 3
}
print(f"字典：{dictionary_0}")

# 直接修改原字典
dictionary_0.clear()
print(f"清空字典：{dictionary_0}")
```
